```python

# Attach competition information to every match
match_data = matches.merge(
    competitions[["comp_id", "comp_name", "match_format", "season"]],
    on="comp_id",
    how="left",
    validate="many_to_one"
)

# Keep only T20 and T20I competitions
match_data = match_data[
    match_data["match_format"].isin(["T20", "T20I"])
].copy()

# Identify IPL separately
match_data["competition_group"] = np.where(
    match_data["comp_name"].str.contains("IPL", case=False, na=False),
    "IPL",
    match_data["match_format"]
)

# Exclude abandoned and no-result matches
result_text = match_data["result"].fillna("").str.lower()

match_data["valid_result"] = ~(
    result_text.str.contains("no result") |
    result_text.str.contains("abandoned")
)

# Check which matches have ball-by-ball data
ball_match_ids = set(ball["match_id"].unique())

match_data["has_ball_data"] = match_data["match_id"].isin(
    ball_match_ids
)

# Check whether each match has a second innings
innings_counts = (
    ball.groupby("match_id")["innings"]
    .agg(lambda x: set(x.dropna().unique()))
)

match_data["has_second_innings"] = (
    match_data["match_id"]
    .map(innings_counts)
    .apply(lambda x: isinstance(x, set) and 2 in x)
)

# Keep valid matches with ball data and a second innings
analysis_matches = match_data[
    match_data["valid_result"] &
    match_data["has_ball_data"] &
    match_data["has_second_innings"]
].copy()

print("MATCH SELECTION SUMMARY")
print("All T20/T20I matches:", len(match_data))
print("Valid-result matches:", match_data["valid_result"].sum())
print("Matches with ball data:", match_data["has_ball_data"].sum())
print("Matches with second innings:", match_data["has_second_innings"].sum())
print("Final eligible matches:", len(analysis_matches))

print("\nMATCHES BY COMPETITION GROUP:")
print(
    analysis_matches.groupby("competition_group")
    .agg(
        matches=("match_id", "nunique"),
        competitions=("comp_name", "nunique"),
        first_season=("season", "min"),
        last_season=("season", "max")
    )
    .to_string()
)

print("\nCOMPETITION-WISE MATCH COUNTS:")
print(
    analysis_matches.groupby(
        ["comp_name", "match_format", "season"]
    )["match_id"]
    .nunique()
    .sort_index()
    .to_string()
)

# Store eligible match IDs for the next step
eligible_match_ids = set(analysis_matches["match_id"])
```

```python

# Keep only deliveries from eligible matches
analysis_ball = ball[
    ball["match_id"].isin(eligible_match_ids)
].copy()

# Normalize extras labels
analysis_ball["extras_lower"] = (
    analysis_ball["extras_type"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
)

# Convert numerical columns safely
numeric_columns = [
    "runs", "extras1_runs", "extras_2_runs"
]

for col in numeric_columns:
    analysis_ball[col] = pd.to_numeric(
        analysis_ball[col], errors="coerce"
    ).fillna(0)

# Total runs added to the team's score
analysis_ball["team_runs"] = (
    analysis_ball["runs"] +
    analysis_ball["extras1_runs"] +
    analysis_ball["extras_2_runs"]
)

# Legal delivery: wides and no-balls are not legal balls
analysis_ball["is_legal"] = ~analysis_ball[
    "extras_lower"
].isin(["wide", "overstep"])

# Calculate innings totals
innings_summary = (
    analysis_ball.groupby(["match_id", "innings"])
    .agg(
        innings_runs=("team_runs", "sum"),
        innings_wickets=("is_wicket", "sum"),
        innings_legal_balls=("is_legal", "sum")
    )
    .reset_index()
)

print("INNINGS SUMMARY:")
print("Rows:", len(innings_summary))
print(innings_summary.head(10).to_string(index=False))

# Extract first-innings totals
first_innings = innings_summary[
    innings_summary["innings"] == 1
][
    ["match_id", "innings_runs", "innings_legal_balls"]
].rename(
    columns={
        "innings_runs": "first_innings_runs",
        "innings_legal_balls": "first_innings_legal_balls"
    }
)

# Extract second-innings totals
second_innings = innings_summary[
    innings_summary["innings"] == 2
][
    ["match_id", "innings_runs", "innings_legal_balls"]
].rename(
    columns={
        "innings_runs": "second_innings_runs",
        "innings_legal_balls": "second_innings_legal_balls"
    }
)

# Build one row per chase
chase_base = (
    analysis_matches.merge(
        first_innings, on="match_id", how="inner"
    )
    .merge(
        second_innings, on="match_id", how="inner"
    )
)

# Use revised target when available; otherwise first innings total + 1
chase_base["target"] = chase_base["revised_target"].fillna(
    chase_base["first_innings_runs"] + 1
)

# Revised second-innings overs field is already expressed in balls
chase_base["max_legal_balls"] = (
    chase_base["second_inn_revised_overs"].fillna(120)
)

print("\nCHASE BASE SUMMARY")
print("Chases:", len(chase_base))
print("Missing targets:", chase_base["target"].isna().sum())
print("Missing ball limits:", chase_base["max_legal_balls"].isna().sum())

print("\nBALL LIMIT DISTRIBUTION:")
print(chase_base["max_legal_balls"].value_counts().sort_index().to_string())

print("\nSAMPLE CHASES:")
print(
    chase_base[
        ["match_id", "comp_name", "first_innings_runs",
         "target", "max_legal_balls", "second_innings_runs"]
    ].head(12).to_string(index=False)
)
```

```python

# Find innings with more legal balls than their permitted limit
innings_check = innings_summary.merge(
    chase_base[
        ["match_id", "comp_name", "max_legal_balls"]
    ],
    on="match_id",
    how="inner"
)

# First innings normally has a 120-ball limit.
# Second innings uses the match-specific revised limit.
innings_check["expected_ball_limit"] = np.where(
    innings_check["innings"] == 1,
    120,
    innings_check["max_legal_balls"]
)

excess_innings = innings_check[
    innings_check["innings_legal_balls"] >
    innings_check["expected_ball_limit"]
].copy()

print("INNINGS EXCEEDING EXPECTED BALL LIMIT:")
print("Count:", len(excess_innings))

print(
    excess_innings[
        ["match_id", "comp_name", "innings",
         "innings_legal_balls", "expected_ball_limit"]
    ].head(30).to_string(index=False)
)

# Inspect the actual deliveries in the first few problematic innings
for match_id in excess_innings["match_id"].head(5):
    for innings_no in excess_innings.loc[
        excess_innings["match_id"] == match_id, "innings"
    ]:
        sample = analysis_ball[
            (analysis_ball["match_id"] == match_id) &
            (analysis_ball["innings"] == innings_no)
        ].copy()

        print(f"\nMATCH {match_id} | INNINGS {innings_no}")
        print("Total deliveries:", len(sample))
        print("Extras:")
        print(sample["extras_lower"].value_counts(dropna=False).to_string())

        print("\nLast 12 deliveries:")
        print(
            sample[
                ["overid", "over_ball_num", "innings_ball_num",
                 "runs", "extras_type", "extras1_runs",
                 "extras_2_runs", "is_wicket", "is_legal"]
            ].tail(12).to_string(index=False)
        )
```

```python

# Summarize innings that exceed 120 legal balls
over_limit_report = (
    analysis_ball.groupby(["match_id", "innings"])
    .agg(
        legal_balls=("is_legal", "sum"),
        max_over_id=("overid", "max"),
        total_deliveries=("match_id", "size")
    )
    .reset_index()
)

over_limit_report = over_limit_report[
    over_limit_report["legal_balls"] > 120
].merge(
    match_data[["match_id", "comp_name", "result"]],
    on="match_id",
    how="left"
)

print("EXCESS INNINGS:", len(over_limit_report))

print("\nLEGAL BALL COUNT DISTRIBUTION:")
print(over_limit_report["legal_balls"].value_counts().sort_index().to_string())

print("\nMAX OVER ID DISTRIBUTION:")
print(over_limit_report["max_over_id"].value_counts().sort_index().to_string())

print("\nEXAMPLES:")
print(
    over_limit_report[
        ["match_id", "comp_name", "innings", "legal_balls",
         "max_over_id", "total_deliveries", "result"]
    ].head(20).to_string(index=False)
)

print("\nINNINGS WITH OVER ID ABOVE 19:")
print(
    over_limit_report[
        over_limit_report["max_over_id"] > 19
    ][
        ["match_id", "comp_name", "innings", "legal_balls",
         "max_over_id", "result"]
    ].head(20).to_string(index=False)
)
```

```python

# Select second innings only
chase_data = analysis_ball[
    (analysis_ball["innings"] == 2) &
    (analysis_ball["match_id"].isin(eligible_match_ids))
].copy()

# Add target and ball limit
chase_data = chase_data.merge(
    chase_base[[
        "match_id", "target", "max_legal_balls", "competition_group"
    ]],
    on="match_id",
    how="inner",
    validate="many_to_one"
)

# Standard T20 maximum: 120 legal balls
chase_data["max_legal_balls"] = (
    chase_data["max_legal_balls"].clip(upper=120)
)

# Order deliveries
chase_data = chase_data.sort_values(
    ["match_id", "innings_ball_num", "overid", "over_ball_num"]
).copy()

# Pre-delivery match state
chase_data["legal_ball"] = chase_data["is_legal"].astype(int)

chase_data["runs_before"] = (
    chase_data.groupby("match_id")["team_runs"].cumsum()
    - chase_data["team_runs"]
)

chase_data["wickets_before"] = (
    chase_data.groupby("match_id")["is_wicket"].cumsum()
    - chase_data["is_wicket"].astype(int)
)

chase_data["legal_balls_before"] = (
    chase_data.groupby("match_id")["legal_ball"].cumsum()
    - chase_data["legal_ball"]
)

# Calculate remaining runs and balls
chase_data["runs_needed"] = (
    chase_data["target"] - chase_data["runs_before"]
).clip(lower=0)

chase_data["balls_remaining"] = (
    chase_data["max_legal_balls"] -
    chase_data["legal_balls_before"]
).clip(lower=0)

# Required run rate
chase_data["required_run_rate"] = np.where(
    chase_data["balls_remaining"] > 0,
    chase_data["runs_needed"] * 6 / chase_data["balls_remaining"],
    np.nan
)

# Keep only deliveries before the target is reached,
# before all 10 wickets fall, and within the ball limit
chase_data = chase_data[
    (chase_data["runs_before"] < chase_data["target"]) &
    (chase_data["wickets_before"] < 10) &
    (chase_data["legal_balls_before"] < chase_data["max_legal_balls"])
].copy()

print("CHASE DELIVERIES:", len(chase_data))
print("MATCHES:", chase_data["match_id"].nunique())
print("PRESSURE STATE SAMPLE:")
print(
    chase_data[[
        "match_id", "striker_name", "runs_before",
        "wickets_before", "legal_balls_before",
        "runs_needed", "balls_remaining", "required_run_rate"
    ]].head(10).to_string(index=False)
)
```

```python

# Apply the agreed pressure definition
chase_data["high_pressure"] = (
    (chase_data["required_run_rate"] >= 12) |
    (
        (chase_data["wickets_before"] >= 6) &
        (chase_data["required_run_rate"] >= 9)
    )
)

# Separate pressure and normal deliveries
pressure_data = chase_data[chase_data["high_pressure"]].copy()
normal_data = chase_data[~chase_data["high_pressure"]].copy()

print("PRESSURE ANALYSIS")
print("Total chase deliveries:", len(chase_data))
print("High-pressure deliveries:", len(pressure_data))
print("Normal deliveries:", len(normal_data))

print(
    "High-pressure percentage:",
    round(len(pressure_data) / len(chase_data) * 100, 2), "%"
)

print("\nPRESSURE MATCHES:", pressure_data["match_id"].nunique())
print("NORMAL MATCHES:", normal_data["match_id"].nunique())

print("\nPRESSURE EXAMPLES:")
print(
    pressure_data[[
        "match_id", "striker_name", "runs_before",
        "wickets_before", "runs_needed",
        "balls_remaining", "required_run_rate"
    ]].head(10).to_string(index=False)
)

# Verify the definition
assert (
    (pressure_data["required_run_rate"] >= 12) |
    (
        (pressure_data["wickets_before"] >= 6) &
        (pressure_data["required_run_rate"] >= 9)
    )
).all()

print("\nPressure definition validation: PASSED")
```

```python

def batting_metrics(df, label):
    data = df.copy()

    # A batter faces a ball on every delivery except a wide
    data["batter_ball"] = (data["extras_lower"] != "wide").astype(int)

    # Count a dismissal only when the striker is the dismissed player
    data["batter_dismissal"] = (
        data["is_wicket"].astype(bool) &
        data["dismissed_player_id"].eq(data["striker_id"])
    ).astype(int)

    result = (
        data.groupby(["striker_id", "striker_name"])
        .agg(
            runs=("runs", "sum"),
            balls=("batter_ball", "sum"),
            dismissals=("batter_dismissal", "sum"),
            matches=("match_id", "nunique")
        )
        .reset_index()
    )

    result[f"{label}_sr"] = np.where(
        result["balls"] > 0,
        result["runs"] / result["balls"] * 100,
        0
    )

    result[f"{label}_avg"] = np.where(
        result["dismissals"] > 0,
        result["runs"] / result["dismissals"],
        np.nan
    )

    return result.rename(columns={
        "runs": f"{label}_runs",
        "balls": f"{label}_balls",
        "dismissals": f"{label}_dismissals",
        "matches": f"{label}_matches"
    })


pressure_batting = batting_metrics(pressure_data, "pressure")
normal_batting = batting_metrics(normal_data, "normal")

batting_comparison = pressure_batting.merge(
    normal_batting,
    on=["striker_id", "striker_name"],
    how="outer"
).fillna(0)

batting_comparison["sr_difference"] = (
    batting_comparison["pressure_sr"] -
    batting_comparison["normal_sr"]
)

print("BATTERS:", len(batting_comparison))
print("Pressure batting records:", len(pressure_batting))
print("Normal batting records:", len(normal_batting))

print("\nTOP 10 BY PRESSURE RUNS:")
print(
    batting_comparison.sort_values(
        "pressure_runs", ascending=False
    )[[
        "striker_name", "pressure_runs", "pressure_balls",
        "pressure_matches", "pressure_sr",
        "pressure_avg", "normal_sr", "sr_difference"
    ]].head(10).round(2).to_string(index=False)
)
```

```python

def batting_metrics_by_format(df, label):
    data = df.copy()
    data["batter_ball"] = (data["extras_lower"] != "wide").astype(int)

    data["batter_dismissal"] = (
        data["is_wicket"].astype(bool) &
        data["dismissed_player_id"].eq(data["striker_id"])
    ).astype(int)

    result = (
        data.groupby(
            ["competition_group", "striker_id", "striker_name"]
        )
        .agg(
            runs=("runs", "sum"),
            balls=("batter_ball", "sum"),
            dismissals=("batter_dismissal", "sum"),
            matches=("match_id", "nunique")
        )
        .reset_index()
    )

    result[f"{label}_sr"] = np.where(
        result["balls"] > 0,
        result["runs"] / result["balls"] * 100,
        0
    )

    result[f"{label}_avg"] = np.where(
        result["dismissals"] > 0,
        result["runs"] / result["dismissals"],
        np.nan
    )

    return result.rename(columns={
        "runs": f"{label}_runs",
        "balls": f"{label}_balls",
        "dismissals": f"{label}_dismissals",
        "matches": f"{label}_matches"
    })


pressure_batting_fmt = batting_metrics_by_format(
    pressure_data, "pressure"
)
normal_batting_fmt = batting_metrics_by_format(
    normal_data, "normal"
)

batting_format_comparison = pressure_batting_fmt.merge(
    normal_batting_fmt,
    on=["competition_group", "striker_id", "striker_name"],
    how="outer"
)

numeric_cols = batting_format_comparison.select_dtypes(
    include="number"
).columns

batting_format_comparison[numeric_cols] = (
    batting_format_comparison[numeric_cols].fillna(0)
)

batting_format_comparison["sr_difference"] = (
    batting_format_comparison["pressure_sr"] -
    batting_format_comparison["normal_sr"]
)

print("FORMAT-WISE BATTER COUNTS:")
print(
    batting_format_comparison.groupby("competition_group")
    .size()
)

print("\nPRESSURE RUN LEADERS BY FORMAT:")
for fmt in ["IPL", "T20", "T20I"]:
    print(f"\n--- {fmt} ---")
    print(
        batting_format_comparison[
            batting_format_comparison["competition_group"] == fmt
        ].sort_values("pressure_runs", ascending=False)[[
            "striker_name", "pressure_runs", "pressure_balls",
            "pressure_matches", "pressure_sr", "normal_sr",
            "sr_difference"
        ]].head(5).round(2).to_string(index=False)
    )
```

```python

def bowling_metrics(df, label):
    data = df.copy()

    # Legal deliveries only count as balls bowled
    data["bowler_legal_ball"] = data["is_legal"].astype(int)

    # Exclude byes, leg-byes and penalty runs from bowler's conceded runs
    non_bowler_extras = data["extras_lower"].isin(
        ["byes", "legbyes", "penalty"]
    )

    data["bowler_runs"] = np.where(
        non_bowler_extras,
        data["team_runs"] - data["extras1_runs"] - data["extras_2_runs"],
        data["team_runs"]
    )

    # Only wickets credited to the bowler
    bowler_wicket_types = {
        "caught", "bowled", "lbw", "stump", "hit wicket"
    }

    data["bowler_wicket"] = (
        data["is_wicket"].fillna(False).astype(bool) &
        data["dismissal_type"].fillna("").str.strip().str.lower()
        .isin(bowler_wicket_types)
    ).astype(int)

    result = (
        data.groupby(["bowler_id", "bowler_name"])
        .agg(
            runs_conceded=("bowler_runs", "sum"),
            legal_balls=("bowler_legal_ball", "sum"),
            wickets=("bowler_wicket", "sum"),
            matches=("match_id", "nunique")
        )
        .reset_index()
    )

    result[f"{label}_economy"] = np.where(
        result["legal_balls"] > 0,
        result["runs_conceded"] * 6 / result["legal_balls"],
        np.nan
    )

    result[f"{label}_sr"] = np.where(
        result["wickets"] > 0,
        result["legal_balls"] / result["wickets"],
        np.nan
    )

    result[f"{label}_avg"] = np.where(
        result["wickets"] > 0,
        result["runs_conceded"] / result["wickets"],
        np.nan
    )

    return result.rename(columns={
        "runs_conceded": f"{label}_runs_conceded",
        "legal_balls": f"{label}_legal_balls",
        "wickets": f"{label}_wickets",
        "matches": f"{label}_matches"
    })


pressure_bowling = bowling_metrics(pressure_data, "pressure")
normal_bowling = bowling_metrics(normal_data, "normal")

bowling_comparison = pressure_bowling.merge(
    normal_bowling,
    on=["bowler_id", "bowler_name"],
    how="outer"
).fillna(0)

bowling_comparison["economy_improvement"] = (
    bowling_comparison["normal_economy"] -
    bowling_comparison["pressure_economy"]
)

print("BOWLERS:", len(bowling_comparison))
print("Pressure bowling records:", len(pressure_bowling))
print("Normal bowling records:", len(normal_bowling))

print("\nTOP 10 BY PRESSURE WICKETS:")
print(
    bowling_comparison.sort_values(
        ["pressure_wickets", "pressure_economy"],
        ascending=[False, True]
    )[[
        "bowler_name", "pressure_wickets",
        "pressure_legal_balls", "pressure_matches",
        "pressure_economy", "pressure_sr",
        "normal_economy", "economy_improvement"
    ]].head(10).round(2).to_string(index=False)
)
```

```python

def bowling_metrics_by_format(df, label):
    data = df.copy()

    non_bowler_extras = data["extras_lower"].isin(
        ["byes", "legbyes", "penalty"]
    )

    data["bowler_runs"] = np.where(
        non_bowler_extras,
        data["team_runs"] - data["extras1_runs"] - data["extras_2_runs"],
        data["team_runs"]
    )

    bowler_wicket_types = {
        "caught", "bowled", "lbw", "stump", "hit wicket"
    }

    data["bowler_wicket"] = (
        data["is_wicket"].fillna(False).astype(bool) &
        data["dismissal_type"].fillna("").str.strip().str.lower()
        .isin(bowler_wicket_types)
    ).astype(int)

    result = (
        data.groupby([
            "competition_group", "bowler_id", "bowler_name"
        ])
        .agg(
            runs_conceded=("bowler_runs", "sum"),
            legal_balls=("is_legal", "sum"),
            wickets=("bowler_wicket", "sum"),
            matches=("match_id", "nunique")
        )
        .reset_index()
    )

    result[f"{label}_economy"] = np.where(
        result["legal_balls"] > 0,
        result["runs_conceded"] * 6 / result["legal_balls"],
        np.nan
    )

    result[f"{label}_sr"] = np.where(
        result["wickets"] > 0,
        result["legal_balls"] / result["wickets"],
        np.nan
    )

    result[f"{label}_avg"] = np.where(
        result["wickets"] > 0,
        result["runs_conceded"] / result["wickets"],
        np.nan
    )

    return result.rename(columns={
        "runs_conceded": f"{label}_runs_conceded",
        "legal_balls": f"{label}_legal_balls",
        "wickets": f"{label}_wickets",
        "matches": f"{label}_matches"
    })


pressure_bowling_fmt = bowling_metrics_by_format(
    pressure_data, "pressure"
)

normal_bowling_fmt = bowling_metrics_by_format(
    normal_data, "normal"
)

bowling_format_comparison = pressure_bowling_fmt.merge(
    normal_bowling_fmt,
    on=["competition_group", "bowler_id", "bowler_name"],
    how="outer"
)

numeric_cols = bowling_format_comparison.select_dtypes(
    include="number"
).columns

bowling_format_comparison[numeric_cols] = (
    bowling_format_comparison[numeric_cols].fillna(0)
)

bowling_format_comparison["economy_improvement"] = (
    bowling_format_comparison["normal_economy"] -
    bowling_format_comparison["pressure_economy"]
)

for fmt in ["IPL", "T20", "T20I"]:
    print(f"\n--- {fmt}: PRESSURE WICKET LEADERS ---")
    print(
        bowling_format_comparison[
            bowling_format_comparison["competition_group"] == fmt
        ].sort_values("pressure_wickets", ascending=False)[[
            "bowler_name", "pressure_wickets",
            "pressure_legal_balls", "pressure_matches",
            "pressure_economy", "pressure_sr",
            "normal_economy", "economy_improvement"
        ]].head(10).round(2).to_string(index=False)
    )

```
